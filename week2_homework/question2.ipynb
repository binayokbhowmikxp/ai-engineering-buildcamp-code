# Question 2: Analyzing Search Results

Search for **capybara** and count titles in the first returned batch that contain `capybara`, ignoring case. The function below uses the same search logic as Question 1.

```python
import requests

def search_wikipedia(query: str):
    response = requests.get(
        "https://en.wikipedia.org/w/api.php",
        params={
            "action": "query",
            "format": "json",
            "list": "search",
            "srsearch": query,
        },
        headers={
            "User-Agent": "ai-engineering-buildcamp/1.0 (educational project)"
        },
        timeout=30,
    )
    response.raise_for_status()
    return response.json()
```

```python
results = search_wikipedia("capybara")
titles = [result["title"] for result in results["query"]["search"]]

count = 0
for title in titles:
    matches = "capybara" in title.lower()
    print(f"{title} — matches: {matches}")
    if matches:
        count += 1

print(f"\nFinal answer: choose {count} titles containing capybara.")
```
