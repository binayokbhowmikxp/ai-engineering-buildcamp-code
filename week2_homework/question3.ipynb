# Question 3: Get Wikipedia Page Content

Fetch the raw wikitext for the exact page title **Capybara** and count its characters with `len(content)`. The count may change when the Wikipedia page is edited.

```python
import requests

def get_page(title: str) -> str:
    response = requests.get(
        "https://en.wikipedia.org/w/index.php",
        params={
            "title": title,
            "action": "raw",
        },
        headers={
            "User-Agent": "ai-engineering-buildcamp/1.0 (educational project)"
        },
        timeout=30,
    )
    response.raise_for_status()
    return response.text
```

```python
content = get_page("Capybara")
content

print("Number of characters:", len(content))

# Preview the beginning of the raw page content.
print("\nPreview:\n", content[:300])
```
