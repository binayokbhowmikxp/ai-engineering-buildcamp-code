# Question 1: Search Wikipedia

Search for **capybara** and display the first batch of results. Count the entries in `query.search` to show how many results this batch contains.

```python
import requests

def search_wikipedia(query: str):
    response = requests.get(
        "https://en.wikipedia.org/w/api.php",
        params={
            "action": "query",
            "format": "json",
            "list": "search",
            "srsearch": query,
        },
        headers={
            "User-Agent": "ai-engineering-buildcamp/1.0 (educational project)"
        },
        timeout=30,
    )
    response.raise_for_status()
    return response.json()
```

```python
results = search_wikipedia("capybara")
search_results = results["query"]["search"]

# Count only the results returned in the first batch.
print("Results in the first batch:", len(search_results))
search_results
```
